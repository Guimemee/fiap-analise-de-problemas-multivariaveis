**Título:** Integrais de Linha e Teorema de Green no Plano

> **Enunciado Original:**
> Calcular o trabalho realizado pelo campo de forças F(x, y) = (y^2, 3xy) ao longo da curva fechada C orientada no sentido anti-horário em torno do retângulo unitário [0, 2] x [0, 3] aplicando o Teorema de Green.

### Formulação Matemática Formal
$$\oint_C (P \, dx + Q \, dy) = \iint_D \left( \frac{\partial Q}{\partial x} - \frac{\partial P}{\partial y} \right) dA = \iint_D (3y - 2y) \, dA = 9$$


### Passo 1: Identificação dos Componentes do Campo Vetorial
P(x, y) = y^2 \quad \text{e} \quad Q(x, y) = 3xy.


### Passo 2: Cálculo do Rotacional Escalar Bidimensional
\frac{\partial Q}{\partial x} = 3y \quad \text{e} \quad \frac{\partial P}{\partial y} = 2y \implies \frac{\partial Q}{\partial x} - \frac{\partial P}{\partial y} = 3y - 2y = y.


### Passo 3: Aplicação do Teorema de Green
\oint_C F \cdot dr = \int_0^2 \int_0^3 y \, dy \, dx.


### Passo 4: Integração na Variável y
\int_0^3 y \, dy = \left[ \frac{y^2}{2} \right]_0^3 = \frac{9}{2}.


### Passo 5: Integração na Variável x e Resultado
W = \int_0^2 \frac{9}{2} \, dx = \frac{9}{2} \times 2 = 9 \text{ Joules de trabalho líquido}.


```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#091a14')
ax.set_facecolor('#091a14')
rect = plt.Rectangle((0, 0), 2, 3, fill=True, color='#122a21', edgecolor='#10b981', lw=2.5)
ax.add_patch(rect)
x = np.linspace(-0.5, 2.5, 12)
y = np.linspace(-0.5, 3.5, 12)
X, Y = np.meshgrid(x, y)
U = Y**2
V = 3*X*Y
ax.quiver(X, Y, U, V, color='#06b6d4', alpha=0.7)
ax.set_xlim(-0.5, 2.5)
ax.set_ylim(-0.5, 3.5)
ax.set_title('Teorema de Green: Curva Fechada e Campo F', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Coordenada x', color='#94a3b8')
ax.set_ylabel('Coordenada y', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#1b4033', alpha=0.4)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
