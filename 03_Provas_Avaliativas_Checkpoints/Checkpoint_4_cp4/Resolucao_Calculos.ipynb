**Título:** Otimização com Vínculos: Multiplicadores de Lagrange

> **Enunciado Original:**
> Determinar as dimensões de um reservatório cilíndrico de volume fixo V = 1.000 m^3 que minimiza a área superficial total de material usando os Multiplicadores de Lagrange.

### Formulação Matemática Formal
$$\nabla A(r, h) = \lambda \nabla V(r, h), \quad h = 2r$$


### Passo 1: Funções Objetivo e Vínculo
Objetivo: A(r, h) = 2\pi r^2 + 2\pi rh. Vínculo: g(r, h) = \pi r^2 h - V_0 = 0.


### Passo 2: Sistema de Equações de Lagrange
\nabla A = (4\pi r + 2\pi h, 2\pi r) \quad \text{e} \quad \lambda \nabla g = \lambda(2\pi rh, \pi r^2).


### Passo 3: Igualdade Componente a Componente
1) 4\pi r + 2\pi h = 2\lambda\pi rh \implies 2r + h = \lambda rh; \quad 2) 2\pi r = \lambda\pi r^2 \implies \lambda = \frac{2}{r}.


### Passo 4: Substituição e Dedução Geométrica
Substituindo \lambda na 1ª equação: 2r + h = \left(\frac{2}{r}\right)rh = 2h \implies h = 2r (altura igual ao diâmetro).


### Passo 5: Solução Numérica do Raio Ótimo
V_0 = \pi r^2(2r) = 2\pi r^3 = 1000 \implies r = \sqrt[3]{\frac{500}{\pi}} \approx 5,42 \text{ m e } h \approx 10,84 \text{ m}.


```python
import numpy as np
import matplotlib.pyplot as plt

r = np.linspace(2, 10, 100)
h = 1000 / (np.pi * r**2)
area = 2 * np.pi * r**2 + 2 * np.pi * r * h
r_opt = (500 / np.pi)**(1/3)
area_opt = 2 * np.pi * r_opt**2 + 2 * np.pi * r_opt * (2*r_opt)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#091a14')
ax.set_facecolor('#091a14')
ax.plot(r, area, color='#10b981', lw=2.5, label='Área Superficial Total (m²)')
ax.axvline(r_opt, color='#ef4444', linestyle='--', label=f'Raio Ótimo r = {r_opt:.2f} m')
ax.scatter([r_opt], [area_opt], color='#ef4444', s=70, zorder=5)
ax.set_title('Otimização com Multiplicadores de Lagrange', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Raio do Cilindro r (m)', color='#94a3b8')
ax.set_ylabel('Área Superficial A (m²)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#1b4033', alpha=0.5)
ax.legend(facecolor='#122a21', edgecolor='#10b981', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
