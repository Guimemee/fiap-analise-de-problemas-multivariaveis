**Título:** Integrais Duplas em Coordenadas Polares

> **Enunciado Original:**
> Calcular o volume de um sólido limitado superiormente pelo paraboloide z = 9 - x^2 - y^2 e inferiormente pelo plano z = 0 usando integral dupla em coordenadas polares.

### Formulação Matemática Formal
$$V = \iint_R (9 - x^2 - y^2) \, dA = \int_0^{2\pi} \int_0^3 (9 - r^2) r \, dr \, d\theta = \frac{81\pi}{2}$$


### Passo 1: Mudança de Variáveis para Coordenadas Polares
Definimos x = r\cos\theta, y = r\sin\theta, portanto x^2 + y^2 = r^2 e o elemento diferencial de área dA = r \, dr \, d\theta.


### Passo 2: Determinação dos Limites de Integração
Para z = 0, temos 9 - r^2 = 0 \implies r = 3. Logo, r \in [0, 3] e \theta \in [0, 2\pi].


### Passo 3: Montagem da Integral Iterada
V = \int_0^{2\pi} \left[ \int_0^3 (9r - r^3) \, dr \right] d\theta.


### Passo 4: Resolução da Integral Radial
\int_0^3 (9r - r^3) \, dr = \left[ \frac{9r^2}{2} - \frac{r^4}{4} \right]_0^3 = \frac{81}{2} - \frac{81}{4} = \frac{81}{4}.


### Passo 5: Resolução da Integral Angular e Resultado Final
V = \int_0^{2\pi} \frac{81}{4} \, d\theta = \frac{81}{4}(2\pi) = \frac{81\pi}{2} \approx 127,23 \text{ unidades de volume}.


```python
import numpy as np
import matplotlib.pyplot as plt

r = np.linspace(0, 3, 50)
theta = np.linspace(0, 2*np.pi, 50)
R, THETA = np.meshgrid(r, theta)
X = R * np.cos(THETA)
Y = R * np.sin(THETA)
Z = 9 - R**2
fig = plt.subplots(figsize=(6, 3.5), facecolor='#091a14')[0]
ax = fig.add_subplot(111, projection='3d')
ax.set_facecolor('#091a14')
ax.plot_surface(X, Y, Z, cmap='plasma', alpha=0.85)
ax.set_title('Sólido do Paraboloide: Integral Dupla', color='#ffffff', fontsize=11, fontweight='bold')
ax.tick_params(colors='#94a3b8')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
