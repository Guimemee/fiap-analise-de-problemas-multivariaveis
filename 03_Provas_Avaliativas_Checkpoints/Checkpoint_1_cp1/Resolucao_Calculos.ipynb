**Título:** Integração por Substituição Trigonométrica e Frações Parciais

> **Enunciado Original:**
> Calcular a integral indefinida de fluxo de calor e campo magnético utilizando a técnica de substituição trigonométrica para expressões da forma sqrt(a^2 - u^2).

### Formulação Matemática Formal
$$\int \frac{dx}{\sqrt{a^2 - x^2}} = \arcsin\left(\frac{x}{a}\right) + C, \quad x = a \sin\theta$$


### Passo 1: Identificação da Substituição Trigonométrica
Para a expressão \sqrt{a^2 - x^2}, definimos x = a \sin\theta, com -\frac{\pi}{2} \le \theta \le \frac{\pi}{2}.


### Passo 2: Diferenciação da Variável Auxiliar
O diferencial dx = a \cos\theta \, d\theta.


### Passo 3: Substituição no Integrando
\sqrt{a^2 - x^2} = \sqrt{a^2 - a^2\sin^2\theta} = a\sqrt{1 - \sin^2\theta} = a \cos\theta.


### Passo 4: Integração na Variável Angular
\int \frac{a \cos\theta \, d\theta}{a \cos\theta} = \int d\theta = \theta + C.


### Passo 5: Retorno à Variável Original
Como \sin\theta = \frac{x}{a}, temos \theta = \arcsin\left(\frac{x}{a}\right). Portanto, o resultado formal é \arcsin\left(\frac{x}{a}\right) + C.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-0.99, 0.99, 200)
y = np.arcsin(x)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#091a14')
ax.set_facecolor('#091a14')
ax.plot(x, y, color='#10b981', lw=2.5, label='f(x) = arcsin(x)')
ax.set_title('Função Primitiva: Substituição Trigonométrica', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Coordenada x', color='#94a3b8')
ax.set_ylabel('Primitiva F(x)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#1b4033', alpha=0.5)
ax.legend(facecolor='#122a21', edgecolor='#10b981', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
