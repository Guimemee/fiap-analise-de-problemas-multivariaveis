**Título:** Vetor Gradiente e Derivada Direcional

> **Enunciado Original:**
> Calcular o vetor gradiente da função de potencial f(x, y) = 2x^2 + 3y^2 no ponto P(1, 2) e determinar a derivada direcional na direção do vetor unitário u que maximiza a taxa de variação.

### Formulação Matemática Formal
$$\nabla f(x, y) = \left( \frac{\partial f}{\partial x}, \frac{\partial f}{\partial y} \right) = (4x, 6y), \quad D_u f = \|\nabla f\|$$


### Passo 1: Determinação do Vetor Gradiente Geral
\nabla f(x, y) = (4x)\mathbf{i} + (6y)\mathbf{j}.


### Passo 2: Avaliação no Ponto P(1, 2)
\nabla f(1, 2) = (4 \cdot 1)\mathbf{i} + (6 \cdot 2)\mathbf{j} = 4\mathbf{i} + 12\mathbf{j}.


### Passo 3: Módulo do Gradiente
\|\nabla f(1, 2)\| = \sqrt{4^2 + 12^2} = \sqrt{16 + 144} = \sqrt{160} = 4\sqrt{10} \approx 12,65.


### Passo 4: Vetor Direcional de Máxima Variação
\mathbf{u} = \frac{\nabla f}{\|\nabla f\|} = \left(\frac{4}{\sqrt{160}}, \frac{12}{\sqrt{160}}\right) = \left(\frac{1}{\sqrt{10}}, \frac{3}{\sqrt{10}}\right).


### Passo 5: Conclusão Física
A taxa máxima de crescimento no ponto P(1,2) ocorre na direção do gradiente e possui intensidade exata de 4\sqrt{10} unidades.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 30)
y = np.linspace(-3, 3, 30)
X, Y = np.meshgrid(x, y)
U = 4*X
V = 6*Y
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#091a14')
ax.set_facecolor('#091a14')
ax.contour(X, Y, 2*X**2 + 3*Y**2, levels=10, colors='#1b4033')
ax.quiver(X[::3, ::3], Y[::3, ::3], U[::3, ::3], V[::3, ::3], color='#10b981', scale=150)
ax.scatter([1], [2], color='#ef4444', s=60, zorder=5, label='Ponto P(1,2)')
ax.set_title('Campo Vetorial Gradiente ∇f e Linhas de Nível', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Coordenada x', color='#94a3b8')
ax.set_ylabel('Coordenada y', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.legend(facecolor='#122a21', edgecolor='#10b981', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
